```python
import os
from dotenv import load_dotenv
load_dotenv()
OPENROUTER_API_KEY=os.environ["OPENROUTER_API_KEY"]
```

```python
from pathlib import Path

import requests

def generate_speech(
    text: str,
    output_path: str = "draft/outputs/generated.mp3",
    model: str = "mistralai/voxtral-mini-tts-2603",
    voice: str = "en_paul_neutral",
)-> Path:
    """
    Generate speech from text using Openrouter.
    """
    response = requests.post(
        "https://openrouter.ai/api/v1/audio/speech",
        headers={
            "Authorization": f"Bearer {OPENROUTER_API_KEY}",
            "Content-Type": "application/json",
        },
        json={
            "model": model,
            "input": text,
            "voice": voice,
            "response_format": "mp3"
        },
        timeout=360,
    )

    response.raise_for_status()

    path = Path(output_path)
    path.parent.mkdir(parents=True, exist_ok=True)
    path.write_bytes(response.content)


```

```python
audio_path = generate_speech(
    text="OpenRouter 官方文档明确说明，这个接口返回的是 raw audio byte stream，也就是原始音频字节流，而非 JSON",
    output_path="outputs/welcome.mp3",
)

print(audio_path)
```
